```python
j = np.int64(1)
k = 1

print('n j k \t\t\t\t\t\t\t\t  binary')
print("-" * 80)

for n in range(1, 66): 
    j = 2*j
    k = 2*k
    print(n, j, k, format(j, "64b"))
```

Every time we multiply by 2 in binary, the computer shifts the register by one digit, effectively overflowing to the "sign" or parity bit, causing the 63rd output to give you a negative number. This is because,

if the parity bit is 1, the computer performs the following:
- use binary complement and flip 0s and 1s
- evaluate num
- add 1
- integer = (-1)*(num+1) 
    
So in an 8-bit system, 10000000 instead of being treated as -0, becomes 11111111, which is interpreted as -128, which is an additional number that would not otherwise be possible to represent in 8-bits.

However, this problem does not arise if we do not limit ourselves to 64-bit representation (as shown by our operation on $k$). By using float point representation of multiple bits, the computer can handle much larger numbers than $2^{63}$.

```python
# Binary to Integer
def Decimal(binary, power=0): 
    if binary == 0:
        return 0
    return (binary % 10) * (2 ** power) + Decimal(binary // 10, power + 1)

num = int(input('Enter binary: '))
print(f'{num} as an integer =', Decimal(num))
```

This recursive program converts each digit in binary to decimal representation by multiplying it with the respective power of 2 and raising the power with each iteration. This logic can be used to convert any integer to an arbitrary bases by simply replacing base 10 with a variable and selecting the appropriate input. 

The function below not only gives us an integer type output, but also lets us convert an integer to any arbitrary base: 

```python
# Integers to arbitrary base
def Convert(x, power=0):

    # Ends recursion
    if x == 0:  
        return 0

    # Finds last digit in the new base, multiplies with corresponding power and recursively adds the output given by the quotient
    return (x % base) * (10 ** power) + Convert(x // base, power + 1)

num = int(input('Enter integer: '))
base = int(input('Enter base: '))
print(f'{num} in base {base} =', Convert(num))
```

In order to convert a decimal to binary, we must deal with the integer and fractional parts separately. The integer parts can be converted in the same way as above. However, the fractional part digits which are in mutiples of $10^{-x}$ must be converted to $2^{-y}$ to be represented in binary. This is achieved by multiplying the decimal fraction by 2, isolating its integer part, and assigning it the respective place value (we use the $10^{-x}$ in the code to assign place value since Python still uses integers). For example, if we want to convert 0.375 to binary: 

$0.375\times2=0.750 \implies int(0.750) = 0 \implies$ Binary = $0.0$ \
$0.750\times2=1.500 \implies int(1.500) = 1 \implies$ Binary = $0.01$ \
$0.500\times2=1.000 \implies int(1.000) = 1 \implies$ Binary = $0.011$ 

This is correct since $0.375 = 2^{-4}+2^{-8}$. Note however, that for certain strings of digits this recursion may take very long to terminate. Hence, we set a manual "precision" to specify the maximum number of decimal places we want our converted fraction to be in the new base.

```python
# Float Decimal to Binary

def DecToBin(num, precision=50):
    
    # Separate integer and fractional parts
    int_part = int(num)
    frac_part = num - int_part

    # Convert integer part to binary
    def convert_int(x, power=0):
        if x == 0:
            return 0
        return (x % 2) * (10 ** power) + convert_int(x // 2, power + 1)
    
    # Convert fractional part to binary
    def convert_frac(y, eps):
        if y == 0:
            return 0
        binary_frac = 0.0
        count = 1
        while y > 0 and count < eps:
            y *= 2
            binary_frac += int(y) * 10**(-count)
            y -= int(y)
            count +=1
        return binary_frac
    
    bin_int = convert_int(int_part)
    bin_frac = convert_frac(frac_part, precision)
    
    return bin_int + bin_frac

num = float(input('Enter decimal: '))
print(f'{num} in binary =', DecToBin(num))
```

We can modify this code to a generalised base by replacing 2 with a variable:

```python
# Float Decimal to Arbitrary Base

def DecToBase(num, precision=10):
    
    # Separate integer and fractional parts
    int_part = int(num)
    frac_part = num - int_part

    # Convert integer part to base
    def convert_int(x, power=0):
        if x == 0:
            return 0
        return (x % base) * (10 ** power) + convert_int(x // base, power + 1)
    
    # Convert fractional part to base
    def convert_frac(y, eps):
        if y == 0:
            return 0
        base_frac = 0.0
        count = 1
        while y > 0 and count <= eps:
            y *= base
            base_frac += int(y) * 10**(-count)
            y -= int(y)
            count +=1
        return base_frac
    
    base_int = convert_int(int_part)
    base_frac = convert_frac(frac_part, precision)
    
    return base_int + base_frac

num = float(input('Enter decimal: '))
base = int(input('Enter base: '))
print(f'{num} in base {base} =', DecToBase(num))
```

### Computational Limitations

## 2.1 Round-Off Problems (Floating-Point Errors)

```python
# Round-off problems
print(1.101 + 1.1 - 2.2)
```

For the above input we expect the output to be 0.001, however we get 0.0009999999999998899 instead. This is a result of the computer converting all values to floating-point numbers in binary. As we saw in the above program, when we convert a decimal to a binary number, the fractional part may be significantly longer in binary representation. When this string exceeds to storage capacity in 64-bit representation, it is truncated, causing a small error in the value. For instance, 

$1.101 \approx 1.0001100110011001100110011001100110011001100110011$ \
$1.1 \approx 1.0001100110011001100110011001100110011001100110011$ \
$2.2 \approx 10.001100110011001100110011001100110011001100110011$ 

Performing the required operations on these approximate binary representations and converting them back to decimals causes the rounding errors to be magnified, producing the observed result.

We can deal with this error by rounding our result to the desired degree of precision that we want. Let's say we are only interested in the result upto 5 decimal places:

```python
print(np.round(1.101 + 1.1 - 2.2, 5))
```

```python
x = 0.3
y = 0.1*3 # floating-point precision in 0.1 is magnified by muliplication 

print(x, y)
print(y-x, np.round(y-x, 5)) # rounding ignores the floating-point error
```

```python
x = np.sqrt(2)
y = 1+(x*1e-14) # Loss of significant digits
z = (y-1)*1e14 # Truncation while subtracting

print(1, y)
print(x, z)
```

$\sqrt{2}$ is an irrational number but the system can only store numbers in finite-precision floating-point numbers. Furthermore, by multiplying with $10^{-14}$, there is a loss of significant digits. During subtraction, the computer shifts the smaller number until both numbers have the same order of magnitude before subtracting. This may cause some digits to overflow from the 64-bit limit, truncating the precision of the computation. 

Hence, you should always avoid subtracting numbers of vastly different orders of magnitude (or be wary of the degree of precision you consider). 

```python
x = np.sqrt(-2) # Irrational number cannot be stored exactly due to finite precision

print(x)
if(x!=x):
    print(f'not equal')
```

## 2.2 Quadratic Roots

```python
# Using the ordinary quadratic formula
b = c = 1

for k in range(2, 64):
    
    a = 10**(-k)
    disc = np.sqrt(b**2 - 4*a*c)
    
    x_plus = (-b + disc)/(2*a)
    x_minus = (-b - disc)/(2*a)
    
    print(a, x_plus, x_minus)
```

We find that as $a\rightarrow0$, the discriminant $\sqrt{b^2-4ac}\rightarrow b$ which causes one of the roots ($x_+ = (-b+\sqrt{b^2-4ac})/2c$) to become unstable since it tends to $0/0$ form. This blows up the floating-point errors and produces unreliable results beyond $a\approx10^{-17}$. This can be dealt with by using the alternate forms for quadratic roots: 

$x_+ = (2c)/(-b-\sqrt{b^2-4ac})$\
$x_- = (2c)/(-b+\sqrt{b^2-4ac})$

Note that the second root is now unstable and blows up to infinity as $a\rightarrow0$. Hence, we should only use the non-problematic roots for both $x_+$ and $x_-$: 

$x_+ = (2c)/(-b-\sqrt{b^2-4ac})$ \
$x_- = (-b-\sqrt{b^2-4ac})/2c$

These will maintain accuracy for much smaller values of $a$ without producing $0$ or $\infty$ for either root.

```python
b = c = 1

for k in range(2, 64):
    
    a = 10**(-k)
    disc = np.sqrt(b**2 - 4*a*c)

    
    x_plus = (2*c)/(-b - disc)
    x_minus = (-b - disc)/(2*a)
    
    print(a, x_plus, x_minus)
```

### Machine Parameters, Evaluating Functions, Bisection Method of Root Finding

## 3.1 Machine Parameters

We can determine the machine precision by checking when the computer determines statements to be equal (==), not equal (!=), greater than (>) or less than (<). We determine the smallest and largest epsilon that can be added to 1 to give the same result  and the smallest number which is not zero by using a while statement with the checking commands. Note that the actual value of the result depends on the initial value since our procedure simply divides the number by 2 until the precision is crossed. Hence, we are only interested in the order of magnitude at which this threshold is crossed.

```python
# Smallest eps where (1+eps)!=1
eps_min = 1.0
while (1.0 + eps_min/2) != 1.0: # Check if eps/2 yields a value equal to 1
    eps_min /= 2 

print(f"Smallest Epsilon = {eps_min} \nOrder of Magnitude = 10^{int(np.log10(eps_min))}")
```

```python
# Largest eps where (1+eps)==1
eps_max = 1e-30
while (1.0 + eps_max*2) == 1.0: # Check if eps*2 yields a value not equal to 1
    eps_max *= 2 

print(f"Largest Epsilon = {eps_max} \nOrder of Magnitude = 10^{int(np.log10(eps_max))}")
```

```python
# Smallest FP-number x_min != 0
x_min = 1.0
while x_min/2 != 0: # Check if x_min/2 yields a value equal to 0
    x_min /= 2

print(f"x_min = {x_min} \nOrder of Magnitude = 10^{int(np.log10(x_min))}")
```

## 3.2 Evaluating Functions 

When computing $f(x) = \sqrt{1+x^2}-1$ for $x\rightarrow0$, then $\sqrt{1+x^2} \approx 1$ which means that our result for $f(x)$ will be unreliable due to finite precision errors and loss of significant digits while subtracting two nearly equal numbers. We can deal with this problem by representing our function in an alternate form that does not require subtracting numbers which are nearly equal: 

$\sqrt{1+x^2}-1 = x^2/(\sqrt{1+x^2}+1)$

This form allows to us to compute the same function for smaller values of $x$ since it does not require computing the difference between two nearly equal numbers.

```python
def f1(x):
    return np.sqrt(1 + x**2) - 1 # Original function

def f2(x):
    return x**2 / (np.sqrt(1 + x**2) + 1) # Stable function

print(" x\t f1(x) \t\t\t f2(x)")
print("-" * 50)

for i in range(1, 20):
    F1 = f1(10**-i)
    F2 = f2(10**-i)
    print(f'{10**-i:.1e}, {F1:.15e}, {F2:.15e}')
```

We observe that the stable function allows us to compute $f(x)$ for values of $x$ well below $10^{-8}$, unlike the original function.

We encounter a similar problem while computing $f(z) = (1-\cos(z))/\sin(z)$ which gives us an indeterminate $0/0$ form for $z\rightarrow0$. We can resolve this using L'Hôpital's rule: 

$f(x) = \lim_{z\rightarrow0} \  (1-\cos(z))/\sin(z) = \tan(z)$

This new function is well defined for all values of $z$.

```python
def f1(z):
    return (1-np.cos(z))/np.sin(z)

def f2(z):
    return np.tan(z) # Stable function

print(" x\t f1(x) \t\t\t f2(x)")
print("-" * 50)

for i in range(1, 20):
    F1 = f1(10**-i)
    F2 = f2(10**-i)
    print(f'{10**-i:.1e}, {F1:.15e}, {F2:.15e}')
```

Note that this approximation does not hold true when $z$ is large, explaining the difference between the answers when $z=0.1$. To visualise the divergence of the approximation, we can plot the functions: 

```python
# Plotting the L'Hôpital's rule approximation
x = np.linspace(0, 1, 1000)

np.seterr(invalid='ignore')
plt.plot(x, f1(x), label=f'(1-cos(z))/sin(z)')
plt.plot(x, f2(x), label=f'tan(z)')
plt.xscale('log')
plt.xlabel(f'$z$')
plt.ylabel(f'$f(z)$')
plt.legend()
plt.show()
```

## 3.3 Bisection Method

```python
# Bisection Method of Finding Roots
def f(x): 
    return x**2 + x - 1 

def Bisection(f, U, L, eps=1e-6, max_iter=1000):

    # Return error if f(U) and f(L) have the same sign
    if f(U)*f(L) >= 0:
        raise ValueError("Function must have opposite signs at U and L")
    
    while abs(U-L)/2 > eps and max_iter >= 0:
        max_iter -= 1
        t = (U+L)/2
        # If f(t) = 0, t is the root
        if f(t) == 0:
            return print(f'Root = {t} \nIterations = {1000-max_iter}')
        # Check which side f(t) lies on
        if f(t)*f(L) < 0:
            U = t
        else:
            L = t
    return print(f'Root = {t} \nPrecision = {eps} \nIterations = {1000-max_iter}')

upper = float(input('Upper Limit = '))
lower = float(input('Lower Limit = '))
Bisection(f, upper, lower)
```

# Lecture 4 (31/01/25)

### Loop, Recursive and Dynamic Programs (Golden Ratio Powers - $\phi^n$)

The nth power of the golden ratio $\phi=(\sqrt{5}-1)/2$ can be estimated using the reccurance relation:  

$\phi_n = \phi_{n-2} - \phi_{n-1}$

Where $\phi_0 = 1$ and $\phi_1 = (\sqrt{5}-1)/2$. Then we find $\phi_n \approx \phi^n$. Hence, we can use this reccurance relation to estimate $\phi^n$.

## 4.1 Loop Method

```python
# Loop program
n = 30

phi = (np.sqrt(5)-1)/2
phi_0 = 1
phi_1 = phi

print(f'n \t phi_n \t\t\tphi^n \t\t\t|phi_n-phi^n|')
print('-' * 60)

for k in range(2, n+1):
    phi_n = phi_0 - phi_1
    print(f'phi_{k:02}, {phi_n:.15e}, {phi**k:.15e}, {abs(phi_n-phi**k):.15e}')
    phi_0 = phi_1
    phi_1 = phi_n
```

This program fails to compute $\phi_n$ accurately despite the recursive form being mathematically valid. Why?

If we assume $\phi_n = k^n$: \
$k^{n+1} = k^{n-1} - k^n$ \
$k^2 + k^1 + 1 = 0$ \
$k = \pm\frac{\sqrt{5}-1}{2}$

So $k_+=\phi$ and we are interested in $\phi^n$. However, the truncated form of $\phi_n$ can be represented by a linear combination of the positive and negative roots of the recursive equation: 

$\phi_n = Ak_+^n + Bk_-^n$

However, since $|k_-| > 1$, the recursion loop will cause the contribution of the second term to blow up for larger $n$, causing the error to grow for larger n.

## 4.2 Recursive Method

The recurrance relation allows us to use a recursive program to find $\phi_n$

```python
# Recursive program
n = 30
phi = (np.sqrt(5)-1)/2

def gmean(n):
    
    if(n<0):
        print('Sorry, negative n recieved')
        
    if(n==0):
        return 1
        
    if(n==1):
        return phi
    
    return gmean(n-2) - gmean(n-1)

phi_n = gmean(n)
phi_N = phi**n

print(f'n \t phi_n \t\t\tphi^n \t\t\t|phi_n-phi^n|')
print('-' * 60)
print(f'{n:.1e}, {phi_n:.15e}, {phi_N:.15e}, {abs(phi_n-phi_N):.15e}')
```

## 4.3 Dynamic Method

The recursive program calculates gmean(n) for the same n multiple times. To reduce computational time, we can store all computed gmean values in a an array and retrieve them when we arrive at the same n again. 

```python
# Dynamic program

n = 30
phi = (np.sqrt(5)-1)/2
arr = np.zeros(n+1)

def gmean_dyn(n):
    
    if(arr[n] != 0):
        return arr[n]
        
    if(n == 0):
        arr[0] = 1
        return 1
        
    if(n == 1):
        arr[1] = phi
        return arr[1]
        
    arr[n] = gmean_dyn(n-2) - gmean_dyn(n-1)
    return arr[n]

phi_n = gmean_dyn(n)
phi_N = phi**n

print(f'n \t phi_n \t\t\tphi^n \t\t\t|phi_n-phi^n|')
print('-' * 60)
print(f'{n:.1e}, {phi_n:.15e}, {phi_N:.15e}, {abs(phi_n-phi_N):.15e}')
```

# Lecture 5 (05/02/25)

### Evaluating oscillating series

```python
# Generate and sum all terms
def consecutive(x, n):
    term = 1
    sum_all = 1
    
    for i in range(1, n):
        term *= (x/i) 
        sum_all += term
        
    return sum_all

x = -16
n = 1000

print('Series Sum:', consecutive(x, n))
print('Actual Value:', np.exp(x))
print('Error:', abs(np.exp(x)-consecutive(x, n)))
```

```python
# Isolate odd and even terms and then sum
def grouped(x, n):
    term_even = 1
    term_odd = x
    sum_even = 1
    sum_odd = x
    
    for k in range(1, int(n/2)):
        for k in range(1, int(n/2) + 1):  # Run from 1 to n/2 inclusive
            term_even *= (x**2 / (2*k * (2*k-1)))
            term_odd *= (x**2 / (2*k * (2*k+1)))
        
            sum_even += term_even
            sum_odd += term_odd

    return sum_even+sum_odd

x = -16
n = 1000

print('Series Sum:', grouped(x, n))
print('Actual Value:', np.exp(x))
print('Error:', abs(np.exp(x)-grouped(x, n)))
```

The Taylor series expansion of $f(x)=e^x$ has been computed in two different ways - summing all consecutive terms and grouping the positive and negative terms before adding them. We find that summing the consecutive terms gives us a more accurate result. This is because the grouping method computes the difference between two large and nearly equal numbers, which as we have previously seen, produces a floating-point precision error.

### Note:
- Do not subtract large numbers, especially if the difference is expected to be small
- Sum consecutive terms when dealing with oscillatory functions

# Lecture 7 (12/02/25)

### Finding Roots: Newton-Raphson and Iterative Methods

## 7.1 Newton-Raphson Method

The Newton-Raphson method is an iterative procedure to find points closer to the root using the derivative of a function. This can be done until the desired precision specified by $\epsilon$. Additionally, we must set an upper limit $N$ on the number of iterations to avoid indefinite computation.

```python
# Function
def f(x): 
    return x**3 - 1
    
# Derivative of Function
def df(x):
    return 3*x**2

# Newton-Raphson Function
def Newton_Raphson(x_0, N, epsilon):
    root_guess = x_0
    for i in range(N):
        root_new = root_guess - f(root_guess)/df(root_guess)
        n = i+1
        if np.abs(root_new - root_guess) < epsilon:
            break
        else:
            root_guess = root_new
    print('Root = ', root_new, '\nIterations = ', n, '\nError = ', f(root_new))

Newton_Raphson(2000, 50, 1e-5)
```

## 7.2 Iterative Method

The iterative method uses recursive procedure to find the roots of a function $f(x)=0$ by representing it as $g(x)=x$ and applying the following reccurance relation: 

$x_n = g(x_{n-1})$ 

Where we provide an initial guess value $x_0$. We iteratively find $x_n$ until it converges to $g(x_n)$, thereby satisfying our equation for a root. However, as we will see shortly, this value does not always converge towards a root. 

```python
def Iterative(F, G, x_0, N, epsilon):
    root_guess = x_0
    print(f'Iter \t Root \t\t Error')
    print('-'*40)
    for i in range(N):
        root_new = G(root_guess)
        n = i+1
        if np.abs(root_new - root_guess) < epsilon:
            break
        else:
            root_guess = root_new

        print(f'{n:02}\t {root_new:.5e}\t {abs(F(root_new)):.5e}')
```

To find the roots of $x^3-1=0$, we can represent it in two different ways to attempt to find the roots: 

$x = 1/x^2 = g_1(x)$\
or\
$x = 1/\sqrt{x} = g_2(x)$

Though both these equations are equivalent, they produce different results when used to find the roots from the iterative approach.

```python
def f(x):
    return x**3 - 1

def g1(x): 
    return 1/x**2

Iterative(f, g1, 1.0001, 50, 1e-5)
```

```python
def f(x):
    return x**3 - 1
    
def g2(x):
    return 1/np.sqrt(x)
    
Iterative(f, g2, 100, 50, 1e-5)
```

For an arbitrary guess, we find that $g_1(x)$ produces diverging values when applied iteratively. However, $g_2(x)$ converges to the positive real root, $x=1$. We can analyse the convergence or divergence of the function by considering its error at iteration $i+1$: 

$\epsilon_{i+1} = |x_{i+1} - r|$

We know that $x_{i+1} = g(x_i)$ and the root is defined by $g(r) = r$, so: 

$\epsilon_{i+1} = |g(x_{i}) - g(r)|$

The intermediate value theorem says: 

$f'(c) = \frac{f(a)-f(b)}{a-b}$ for $a\leq c \leq b$

So we can say $|g(x_{i}) - g(r)| = |g'(x)||(x_i-r)|$, but $\epsilon_i = |x_i-r|$, so: 

$|g'(x)| = \epsilon_{i+1}/\epsilon_{i}$

Hence, if $|g'(x)|<1$ for $r \leq x \leq x_0$, then the error will reduce with each iteration, thereby eventually converging towards the root. We can now analyse the functions to determine their convergence conditions: 

$|g'_1(x)| = 2/x^3$

This function is convergent only for $x \geq 2^{1/3}$ or $x \leq -2^{1/3}$, which does not contain the real roots $x=1, -1$ and therefore cannot converge to either of them. 

$|g'_2(x)| = 1/2x^{-3/2}$

This function is convergent for $0 \leq x \leq 4^{1/3}$, which contains $x=1$ and therefore only converges to the positive real root. 

### 7.2.1 Example: Find the roots of $f(x)=x^2+x-1$

```python
def f(x):
    return x**2 + x - 1

def f1(x):
    return 1/(x+1)

Iterative(f, f1, -0.01, 100, 1e-5)
```

```python
def f2(x):
    return (1-x)/x

Iterative(f, f2, -0.01, 100, 1e-5)
```

Depending on which function we choose for $g(x)$, we get different convergent conditions which contain different roots.

# Lecture 8 (14/02/25)

### Newton-Raphson Fractral

A Newton-Raphson Fractal is produced by using the Newton-Raphson method for arbitrary points on the complex plane and colour coding the coordinates according to the root to which they converge. For $x^n=1$ we find that there are $n$ roots and therefore $n$ colours in the fractal.

```python
# Newton-Method

n = 3
a = 1

# Function
def f(x): 
    return x**n - 1
    
# Derivative of Function
def df(x):
    return n*x**(n-1)

# Newton-Raphson Method
def newton_method(x0, eps=1e-5, N=50):
    for i in range(N):
        x1 = x0 - a*(f(x0)/df(x0))
        if np.abs(x1 - x0) < eps:
            break
        else:
            x0 = x1
    return x1
```

```python
# Newton Fractal
res = 500
lim = 3
x = np.linspace(-lim, lim, res)
y = np.linspace(-lim, lim, res)
X, Y = np.meshgrid(x, y)
Z = X + Y*1j

R = [np.exp(2j * np.pi * k / n) for k in range(n)] # Roots of x^n=1

fractal = np.zeros(Z.shape, dtype=int)

for i in range(res):
    for j in range(res):
        root = newton_method(Z[i, j])
        err = [np.abs(root - r) for r in R]
        fractal[i, j] = np.argmin(err)

plt.style.use("dark_background")
plt.figure(figsize=(8, 8))
plt.imshow(fractal, extent=(-lim, lim, -lim, lim), cmap='viridis')
plt.title(f'Newtons Fractal for $f(x) = x^{n} - 1, \ a = {a}$')
plt.xlabel('Re(x)')
plt.ylabel('Im(x)')

# Legend (Created using ChatGPT)
colors = [plt.cm.viridis(i / (n - 1)) for i in range(n)]
patches = [mpatches.Patch(color=colors[i], label=f'{np.round(R[i], 3)}') for i in range(n)]
plt.legend(handles=patches, loc='upper left', fontsize=10, frameon=True)

plt.show()
```

### Numerical Integration: Trapezoidal Method and Simpson's 1/3rd Method, Lagrange Interpolation

## 9.1 Trapezoidal Integral

The Trapezoidal Integral numerically finds a definite integral of a function $f(x)$ between $a$ and $b$ by finding the Riemann Sum of the areas of $n$ different trapezoids defined by their common width $dx=(b-a)/n$ and heights $f(a)$ and $f(b)$. 

```python
def f(x):
    return np.exp(x)

def trap_int(f, a, b, eps, N):
    
    I = [] # Store integral values
    
    for i in range(1, N):
        n = 10**i
        x = np.linspace(a, b, n + 1)
        y = f(x)
        dx = (b-a)/n

        I_n = (dx/2) * (2*np.sum(y[1:-1]) + y[0] + y[-1]) # Trapezoidal Integral
        I.append(I_n) 

        if len(I)>1 and abs(I[-1]-I[-2])<eps: # Break when difference is less than eps
            break

    return print(f'Integral = {I[-1]} \nError = {abs(I[-1]-I[-2])} \nIterations = {i}')
        
trap_int(f, -1, 1, 1e-10, 20)
```

## 9.2 Simpson's 1/3rd Integral 

Simpson's 1/3rd Integral numerically finds a definite integral of a function $f(x)$ between $a$ and $b$ by finding the Riemann Sum of $n$ different areas with a width of $dx=(b-a)/n$ and fitting a quadratic function between $f(x_{i-1})$, $f(x_{i})$ and $f(x_{i+1})$. The resulting coefficients are determined analytically before computing the summation. 

```python
def f(x):
    return np.exp(x)

def simp_int(f, a, b, eps, N):
    
    I = [] # Store integral values
    
    for i in range(1, N):
        n = 10**i
        x = np.linspace(a, b, n + 1)
        y = f(x)
        dx = (b-a)/n

        I_n = (dx/3) * (y[0] + y[-1] + 4*np.sum(y[1:-1:2]) + 2*np.sum(y[2:-2:2])) # Simpson Integral
        I.append(I_n) 

        if len(I)>1 and abs(I[-1]-I[-2])<eps: # Break when difference is less than eps
            break

    return print(f'Integral = {I[-1]} \nError = {abs(I[-1]-I[-2])} \nIterations = {i}')
        
simp_int(f, -1, 1, 1e-10, 20)
```
