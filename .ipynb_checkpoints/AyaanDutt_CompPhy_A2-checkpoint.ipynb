# 1. Coupon Collector’s Problem

## 1.1 Picking Cards

In this variation of the coupon collector's problem, we pick cards from a well-shuffled deck until we draw at least one card from each suit. The main difference between this and the general coupon collector's problem is that the probability of the suit of each card is dependent on the previously picked cards. Hence, we need to first generate a finite deck before picking cards. 

```python
def card_draws(num_suits, deck_size):
    deck = np.arange(deck_size) # Generate a deck of 52 cards

    for i in range(deck_size): # Shuffle the deck
        i_new = i + int((deck_size-i)*np.random.rand())
        # swap i and i_new
        temp = deck[i]
        deck[i] = deck[i_new]
        deck[i_new] = temp

    drawn_suits = np.zeros(num_suits, dtype=bool)
    draws = 0
    unique_suits = 0

    for card in deck:
        suit = card % num_suits # Suit given by remainder of card when divided by no. of suits
        draws += 1
        
        if not drawn_suits[suit]: # Store drawn suits 
            drawn_suits[suit] = True
            unique_suits += 1
            
        if unique_suits == num_suits: # If drawn suits = number of suits, all suits have been drawn
            break
            
    return draws # Return total no. of draws

N = 10000 # No. of iterations
all_draws = np.array([card_draws(4, 52) for i in range(N)]) 

avg_draws = np.mean(all_draws)
std_draws = np.std(all_draws)

print(f'Number of Trials = {N} \nAverage Draws = {avg_draws} \nStandard Deviation of Draws = {std_draws}')

plt.hist(all_draws, bins=20, color='#505aae', density=True)
plt.xlabel('Number of Draws')
plt.ylabel('Normalized Probability Density')
plt.show()
```

## 1.2 Coupon Collector

In this problem, each coupon is equally likely and independent of each other. Hence, we can pick the coupon at random each time, without having to worry about the probabilities of the remaining coupons. 

```python
def collector(n, N):
    collection_times = np.zeros(N)
    
    for trials in range(N): 
        collected = np.zeros(n, dtype=bool) 
        count = 0
        unique_coupons = 0
        
        while unique_coupons < n: # Draw coupons until n unique coupons are drawn 
            coupon = int(n*np.random.rand()) # Generate random coupon between 0 to n 
            count += 1
            
            if not collected[coupon]: # Store collected coupons
                collected[coupon] = True
                unique_coupons += 1

        collection_times[trials] = count # Store total no. of draws

    avg_time, std_time = np.mean(collection_times), np.std(collection_times) 

    return collection_times, avg_time, std_time # Return total, avg and std of draws

N = 10000 # No. of iterations
collection_times, avg_time, std_time = collector(4, N)

print(f'Number of Trials = {N} \nAverage Draws = {avg_time} \nStandard Deviation of Draws = {std_time}')

plt.hist(collection_times, color='#505aae', bins=20, density=True)
plt.xlabel('Number of Draws')
plt.ylabel('Normalized Probability Density')
plt.show()
```

Since each pick is independent and equally likely, after picking $i$ types of coupons we expect the probability of a next coupon being a new type to be $(n-i)/n$. On average, the number of draws needed to get a new type of coupon is therefore $n/(n-i)$. Hence, the total number of draws required to get all $n$ coupons will simply be the sum of the number of draws needed to pick a new coupon for every $i=1, 2, 3 ..., n$:

Total collection time = $\displaystyle 1+\frac{n}{n-1}+\frac{n}{n-2}+...+\frac{n}{2}+\frac{n}{1} = \displaystyle n\sum_{i=1}^n \frac{1}{i}$ 

Hence, the average collection time for choosing $n$ different types of coupons where each coupon is equally likely is $n\sum_{i=1}^n (1/i)$.

```python
# Simulate for different n
sim = 30 # No. of simulations
x = np.arange(sim)
avg_times = np.zeros(sim)

for n in x:
    N = 1000
    avg_times[n]= collector(n, N)[1]

# Expected collection time 
def exp_collection_time(x):
    return x*np.array([np.sum(1/np.arange(1, k+1)) for k in x])
```

```python
plt.scatter(x, avg_times, c='#f48a47', label=f'Simulated (N={N})') # Plot simulated average draws
plt.plot(x, exp_collection_time(x), c='#a11b9a', label='Expected') # Plot expected average draws
plt.xlabel('Number of Coupons')
plt.ylabel('Average Waiting Time')
plt.legend()
plt.show()
```

# 2. 2D Random Walk

## 2.1 Lattice Random Walk

A 2D random walk can be simulated by making 2 decisions at every step t: 
1. Take step in x or y direction
2. Take step in +1 or -1 direction (set $r[t+1]=r[t]$ for other direction)

We can store every step along each axis in separate arrays and run the simulation many times. By extracting the final positions for each run, we can determine the probability distribution function vs x & y. 

Using the series of x and y data, we can also plot the mean square displacement $\langle R^2\rangle = \langle x^2+y^2\rangle$ against the corresponding number of steps $N$. 

```python
# 2D Random Walk
def walk_2D(N_steps):
    
    x, y = np.zeros(N_steps), np.zeros(N_steps)

    for t in range(N_steps-1):

        # Move in x-axis
        if np.random.rand() < 0.5: 
            if np.random.rand() < 0.5:
                x[t+1] = x[t] + 1 # Step right
            else:
                x[t+1] = x[t] - 1 # Step left
            y[t+1] = y[t]
            
        # Move in y-axis
        else:                       
            if np.random.rand() < 0.5:
                y[t+1] = y[t] + 1 # Step up   
            else:
                y[t+1] = y[t] - 1 # Step down
            x[t+1] = x[t]

    return x, y
```

By tracing the path of a single random walk and colouring it according to the time of the step, it is possible to observe the behaviour of a random walk. I wrote the code for this as a part of my Brownian motion lab report in Physics Lab 4. 

```python
N_steps = 1000 # Number of steps

x, y = walk_2D(N_steps)

t = np.arange(N_steps)

# Plot Random Walk Path
p = np.array([x, y]).T.reshape(-1, 1, 2)  
s = np.concatenate([p[:-1], p[1:]], axis=1)  

norm = plt.Normalize(t.min(), t.max())

lc = mcoll.LineCollection(s, cmap='plasma', norm=norm, linewidth=2) # Generate colourmap for path
lc.set_array(t[:-1])

plt.gca().set_facecolor((0, 0, 0, 0.3))

plt.gca().add_collection(lc)

start_color = lc.get_array()[0]
end_color = lc.get_array()[-1]

# Mark start and end
plt.scatter(x[0], y[0], color=cm.plasma(norm(start_color)), zorder=5, label='Start')
plt.scatter(x[-1], y[-1], color=cm.plasma(norm(end_color)), zorder=5, label='End')

plt.colorbar(lc, label='Time')
plt.xlabel(r'x-position')
plt.ylabel(r'y-position')
plt.title('Trajectory of a Single Particle')
plt.legend(loc='lower right')
plt.axis('equal')
plt.show()
```

## 2.2 Simulation

```python
# Simulation
def simulate_walks_2D(N_steps, N_walks):
    
    X = np.zeros((N_walks, N_steps))
    Y = np.zeros((N_walks, N_steps))
    final_x = np.zeros(N_walks)
    final_y = np.zeros(N_walks)

    for i in range(N_walks):
        
        x, y = walk_2D(N_steps)
        
        # Store final positions of run
        final_x[i] = x[N_steps-1] 
        final_y[i] = y[N_steps-1]
        
        # Store all positions of run
        X[i] = x
        Y[i] = y

    return final_x, final_y, X, Y
```

```python
N_steps = 1000 # Number of steps per run
N_walks = 10000 # Number of runs simulated

final_x, final_y, X, Y = simulate_walks_2D(N_steps, N_walks)
```

## 2.3 Probability Density Distribution

```python
#%matplotlib widget
# Surface plot of probability density distribution
hist, xedges, yedges = np.histogram2d(final_x, final_y, bins=40, density=True)

P_N = gaussian_filter(hist, sigma=1.0) # Smoothen probability density function 

xcenters = (xedges[:-1] + xedges[1:]) / 2
ycenters = (yedges[:-1] + yedges[1:]) / 2
X_final, Y_final = np.meshgrid(xcenters, ycenters) # Set center of bins

fig = plt.figure(figsize=(6, 6))
ax = fig.add_subplot(111, projection='3d')

surf = ax.plot_surface(X_final, Y_final, P_N.T, cmap='viridis', edgecolor='none', antialiased=True) # Surface plot

ax.set_xlabel(f'Final X Position', fontsize=12)
ax.set_ylabel(f'Final Y Position', fontsize=12)
ax.set_zlabel(f'Probability Density $P_N(R)$', fontsize=12)
ax.set_title(f'Surface Plot of Probability Density Distribution', fontsize=14)
plt.show()
```

The rate function for large deviations for the probability density function is given by: 

$\displaystyle I(R)=-\lim_{N\rightarrow\infty} \frac{1}{N} \ln(P_N(R))$

Hence, we can simply plot a surface plot of $I(R)$ for sufficiently large $N$ (in my case, I have taken $N=10,000$). 

```python
I_R = -(1/N_steps)*np.log(P_N)

fig = plt.figure(figsize=(6, 6))
ax = fig.add_subplot(111, projection='3d')

surf = ax.plot_surface(X_final, Y_final, I_R.T, cmap='viridis', edgecolor='none', antialiased=True) # Surface plot

ax.set_xlabel(f'Final X Position', fontsize=12)
ax.set_ylabel(f'Final Y Position', fontsize=12)
ax.set_zlabel(f'Rate Function $I(R)$', fontsize=12)
ax.set_title(f'Surface Plot of Rate Function for Large Deviations', fontsize=14)
plt.show()
```

## 2.4 Mean Square Displacement

At each step, along each axis the walker moves:
1. $+1$ with a probability of $1/4$
2. $-1$ with a probability of $1/4$
3. $0$ with a probability of $1/2$ (moves in other axis)

Hence, $\langle \Delta x_i^2\rangle = \frac{1}{4}(1)^2+\frac{1}{4}(-1)^2+\frac{1}{2}(0)^2 = \frac{1}{2}$, and similarly, $\langle \Delta y_i^2\rangle = \frac{1}{2}$. For $N$ steps:

$\displaystyle \langle x_N^2\rangle = \sum_{i=1}^N \langle \Delta x_i^2\rangle = \frac{N}{2}$ and $\displaystyle \langle y_N^2\rangle = \frac{N}{2}$

Therefore, the mean square displacement for $N$ steps is: 

$\displaystyle \langle R_N^2\rangle = \langle x_N^2\rangle + \langle y_N^2\rangle = N$

```python
# Mean Square Displacement vs Steps
MSD = np.mean((X**2+Y**2), axis=0)
steps = np.arange(N_steps)

plt.scatter(steps, MSD, s=5, color='#f48a47', label='Simulated')
plt.plot(steps, steps, c='#a11b9a', label='Expected')
plt.xlabel(r'Steps [N]')
plt.ylabel(r'Mean Square Displacement $\langle R_N^2\rangle$')
plt.legend()
plt.show()
```

# 3. Rumors: Jack and Jill

## 3.1 Recursive Rumor Spread

The rules of the rumor spreading through a party with N people (excluding Jill) are:
1. Jack starts the rumor by telling one person (who is not Jill)
2. Each person who hears the rumor for the first time tells it to 1 other guest (excluding the person they heard it from, and Jill).
3. Once someone hears it again, they stop spreading it.
4. The process ends when no new person hears the rumor.

We can program the spread of the rumor at the individual level by using a recursive program, since the process is essentially repeating, with the previous listener becoming the next source. 

```python
def Rumor_Spread(N):
    heard = set([0]) # Jack(=0) starts the rumor

    def Spread(current, source):

        # Potential listeners are all people except source and current
        listeners = np.array([ppl for ppl in range(N) if ppl != source and ppl != current]) 

        i = int(len(listeners)*np.random.rand()) # Pick a random potential listener as the target
        target = listeners[i]

        if target not in heard: # If target has not heard the rumor, add to heard and spread again
            heard.add(target)
            Spread(target, current)

    first = int((N-1)*np.random.rand()) + 1 # Pick random first target (minus Jack himself)
    heard.add(first)
    Spread(first, 0)

    return len(heard), (len(heard) == N) # Return number of people who heard and check if all people heard
```

## 3.2 Monte-Carlo Simulation

We can now run a Monte-Carlo simulation to test how often the rumor successfully spreads through the entire party and what the average number of people who heard the rumor is, given a party with N people.

```python
def Rumor_Simulation(N, trials):
    heard_counts = np.zeros(trials)
    all_heard = 0

    for i in range(trials):
        num_heard, success = Rumor_Spread(N)
        heard_counts[i] = num_heard
        if success:
            all_heard += 1

    prob_heard = all_heard/trials
    exp = np.mean(heard_counts)
    return prob_heard, exp
```

```python
N = 3 # Number of people at party 
trials = 10000 # Number of simulations

P_n, exp = Rumor_Simulation(N, trials)
print(f'No. of people at party = {N}')
print(f'No. of trials = {trials}')
print(f"Probability that all people hear the rumor: {P_n}")
print(f"Expected number of people who hear the rumor: {exp}")
```

## 3.3 Rumor Trends

In order to determine how the expected number of people who hear the rumor changes with number of people at the party, we can run the simulation for various $N$. We can also see how the probability changes for different $N$.

```python
N_max = 100
avg_heard = np.zeros(N_max)
Pn = np.zeros(N_max)

for i in range(N_max):
    trials = 1000
    # Find probability and average number of people who hear the rumor for N people (starting from N=3)
    Pn[i], avg_heard[i] = Rumor_Simulation(i+3, trials) 
```

```python
plt.figure(figsize=(6, 4))
plt.scatter(range(3, N_max+3), avg_heard, s=5, c='teal')
plt.xlabel('Number of People [N]')
plt.ylabel('Average Number of People Who Heard [$N_{exp}$]')
plt.show()
```

```python
plt.figure(figsize=(6, 4))
plt.plot(range(3, N_max+3), Pn, c='teal')
plt.xlabel('Number of People [N]')
plt.ylabel('Probability of 100% Successful Rumor')
plt.xscale('log')
plt.show()
```

Note that the probability of a 100% successful rumor quickly approaches 0 for a party with more than 10 people. However, a larger number of trials would give us a more precise estimate of how the probability declines.

# 4. Volume of a d-Dimensional Sphere

## 4.1 Monte-Carlo Simulation

```python
def Sphere_Volume(d, N):
    
    in_sphere = 0
    
    for p in range(N):
        coordinates = 2*np.random.uniform(size=d)-1 # Randomly generate a point in d-dim space

        if np.sum(coordinates**2) <= 1: # Check if point is inside the sphere
            in_sphere += 1 

    return (in_sphere/N)*2**d # Return volume of sphere
```

## 4.2 $V_d$ vs $d$

For a d-dimensional unit sphere, the volume is found to be (source: Wikipedia): 

$\displaystyle V_d = \frac{\pi^{d/2}}{\Gamma(d/2+1)}$

We can use this to determine the expected volume. To find this, we can use recursion to create a simple gamma function.

```python
# Gamma Function (from ChatGPT)
def gamma(x):
    if x == 1:
        return 1
    elif x == 0.5:
        return np.sqrt(np.pi)
    elif x > 1 and x == int(x):
        return np.prod(np.arange(1, int(x)))  # (x-1)!
    elif x > 0:
        return (x-1)*gamma(x - 1)
```

```python
dim = np.arange(1, 21)
V_d = np.zeros(len(dim))
V_exp = np.zeros(len(dim))

for i in range(len(dim)): # Find the simulated and expected volumes
    V_d[i] = Sphere_Volume(dim[i], 50000)
    V_exp[i] = np.pi**(dim[i]/2)/gamma(dim[i]/2 + 1)
```

```python
plt.scatter(dim, V_d, c='#f48a47', label='Simulated')
plt.plot(dim, V_exp, c='#a11b9a', ms=5, label='Expected')
plt.xlabel(r'Dimension [d]')
plt.ylabel(r'Volume of d-Dimensional Sphere [$V_d$]')
plt.legend()
plt.show()
```

## 4.3 Lagrangian Interpolation

Using the generalised Lagrangian Interpolation formula, we can determine the volume of a sphere of "fractional-dimensions" as well. 

$\displaystyle P(x) = \sum_{i=0}^{n} y_i \cdot \ell_i(x), \quad \text{where} \quad \ell_i(x) = \prod_{\substack{j=0 \\ j \ne i}}^{n} \frac{x - x_j}{x_i - x_j}$

We can write a function that takes the points of our simulation data as an input and uses it to interpolate for $d=5.3$. 

```python
def lagrange_interpolation(x, y, x_0):
    result = 0

    for i in range(len(x)):
        li = 1
        for j in range(len(x)):
            if i != j:
                li *= (x_0 - x[j]) / (x[i] - x[j])
                
        result += y[i] * li

    return result
```

```python
# Using d=(1, 10)
V = lagrange_interpolation(dim[0:11], V_d[0:11], 5.3)

print(f'V_d for (d={5.3}) = {V}')
```

# 5. Birthday Problem 

```python
def shared_birthday(n, trials):
    birthdays = (365*np.random.rand(trials, n)).astype(int) # Generate n birthdays for each trial
    
    counts = np.array([len(np.unique(row)) for row in birthdays]) # Count number of unique birthdays
    
    shared = np.sum(counts < n) # Count number of times birthdays are shared
    
    return shared/trials # Return number of shared birthdays / number of trials
```

```python
N = 100
trials = 10000
P_n = np.zeros(N)

for n in range(N):
    P_n[n] = shared_birthday(n, trials)
```

```python
plt.figure(figsize=(6, 4))
plt.scatter(range(N), P_n, s=5, c='#f48a47', label='Simulated $P_n$')
plt.axhline(0.5, color='#a11b9a', linestyle='--', label='50% threshold')
plt.xlabel('Number of People (n)', fontsize=12)
plt.ylabel('Probability of Shared Birthday ($P_n$)', fontsize=12)
plt.legend()
plt.show()
```

As we expect, the probability of at least 2 people sharing a birthday exceeds 50% for a group of 23 people. After around 50 people, the probability is nearly 100%.
